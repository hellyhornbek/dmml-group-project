# Exploring the cleaned Dataminers data

A first look at `dataminers_cleaner.csv`: how big it is, what is in it, and how the motivations differ between years.

Each section below answers one question. Every line of code has a short comment above it saying what it does.

## Setup

```python
# pandas is the library for working with tables (DataFrames)
import pandas as pd

# Read the cleaned CSV into a DataFrame (the path is relative to this notebook's folder)
df = pd.read_csv("../cleaned/dataminers_cleaner.csv")
```

## 1. Size and first look

**Question:** how many rows and columns are there, and what do the first rows look like?

```python
# .shape gives (number of rows, number of columns), without parentheses since it is an attribute
display(df.shape)
display(df.head(5))
```

## 2. Types and missing values

**Question:** which columns are numbers and which are text, and which columns have missing values?

```python
# .isna() marks each missing cell as True, and .sum() counts the Trues per column
df.isna().sum()
```

```python
# .dtypes shows the data type of each column: int64/float64 are numbers, object is text
display(df.dtypes)
```

## 3. Who is in the data?

**Question:** how many students are there per programme, and per source file (year)?

```python
# value_counts() counts how often each programme occurs, most common first
df.groupby("programme_clean").size().sort_values(ascending=False)
```

```python
# Same for the source file, with sort_index() so the years come in order
df.groupby("source_file").size().sort_values(ascending=False)
```

## 4. The numbers

**Question:** what are the minimum, maximum and mean of `height_cm` and `shoe_eu`?

```python
# Pick the two columns with a list of names, and describe() gives count, mean, std, min, quartiles and max
display(df[["height_cm", "shoe_eu"]].describe())
```

```python
height_min = df["height_cm"].min()
display(df[df["height_cm"] == height_min])
```

## 5. Motivations

**Question:** what share of students ticked each motivation, overall and per year?

Each of the four columns is 0 or 1, so the **mean of a column is the share of students who ticked it**.
The shares do not add up to 100%, because a student can tick more than one reason.

```python
# The four 0/1 motivation columns, collected in one list so they can be reused
motivation_cols = ["exciting", "mandatory", "commercial_skills", "uncertain"]
```

```python
# Overall share of students who ticked each motivation
df[motivation_cols].mean()
```

```python
# Split the rows by year with groupby, then take the mean of the same four columns within each year
df.groupby("source_file")[motivation_cols].mean()
```

## Findings so far

- The data has 222 students and 14 columns, and only `seattle` has missing values (58).
- `mandatory` is about 0.44, 0.42 and 0.33 in 2023-2025 and drops to about 0.04 in 2026.
- `exciting` is higher in 2025 and 2026 than in 2023-2024, and `uncertain` has shrunk over the years.
