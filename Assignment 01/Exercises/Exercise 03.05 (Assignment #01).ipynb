# Exercise 03.05 (Assignment #01) — Solving a System of Linear Equations

```python
import numpy as np
```

## The system of equations

$$2x + 3y = 5$$
$$4x + y = 6$$

Represent this system in matrix form `Ax = b`.

```python
A = [[2, 3],
     [4, 1]]
b = [5, 6]
```

## Solving for x

Use NumPy to solve `Ax = b` for the vector `x = [x, y]`.

```python
result = np.linalg.solve(A, b)
print(f"x = {result[0]}, y = {result[1]}")
```
