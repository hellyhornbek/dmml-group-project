## Interactive scatter plot: height vs mass
1. We filter out the species with 1 character, and are left with 58/87 characters. 
2. Then we calculate the Pearson r for all characters and for all excluding Jabba. 


```python
counts = df["species"].value_counts()
common_species = counts[counts >= 2].index
df_common = df[df["species"].isin(common_species)]

print(f"Kept {len(df_common)} of {len(df)} characters, across {df_common['species'].nunique()} species")

fig_scatter = px.scatter(
    df_common,
    x="mass",
    y="height",
    color="species",
    hover_name="name",
    hover_data=["gender", "homeworld"],
    title="Height vs mass by species (species with 2+ characters)",
    labels={"mass": "Mass (kg)", "height": "Height (cm)"},
)
fig_scatter.show()
```

```python
all_chars = df[["height", "mass"]]
without_jabba = df.loc[df["mass"] < 1000, ["height", "mass"]]

print(f"Pearson r, all characters:  {all_chars['height'].corr(all_chars['mass']):.3f}")
print(f"Pearson r, excluding Jabba: {without_jabba['height'].corr(without_jabba['mass']):.3f}")
```

* Calculating the pearson r for all characters gave 0.11, which is essentially no linear relationship. But by excluding Jabba it rises to 0.65, which is a clear positive linear relationship.
* Humans cluster tighly in the middle of the plot. But this is also partly artificial, as we have median imputed 23 characters mass. Those sit on one visible line.
