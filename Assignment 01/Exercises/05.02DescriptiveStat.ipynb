### 2.  Compute Descriptive Statistics: 
Mean, median, mode and standard dev.

```python
df.describe
df.shape
df[["mass", "height", "birth_year"]].describe()
```

```python
print(df["mass"].skew())
print(df["height"].skew())
print(df["birth_year"].skew())
```

```python
mode = df[["mass", "height", "birth_year"]].mode()
print(mode)
```

* For height: std dev is much smaller than mean, thats good. Indicates that the data points are tightly clustered around the average. Not the case for mass and birth_year
* For "mass" and "birth_year": Because mean is sensitive to outliers, it is much bigger than the median in both columns. This hints that there are outliers. 
* The standard deviation in the same two columns are also larger than the mean, which also indicates that one or more outliers are creating variance. 
* For height: Mean and median are quite close. Median being slightly higher, indicating that there must be some very short characters as outliers. 
* All three are notably skewed. But height is the least skewed and the only one skewed to the left. Means oppositely that a few unusually low values are dragging the average down and the tail is to the left.
* Mode is completely unaffected by outliers, but it is certainly affected by imputation. More on this in the section of missing values. 

### Variance

```python
var = df[["mass", "height", "birth_year"]].var()
print(var)

cv = df[["mass", "height", "birth_year"]].std() / df[["mass", "height", "birth_year"]].mean()
print(cv)
```

 * CV = coffeficient of variance. To compare relative variability and not just std. dev i divited the std deviation with the mean for each to cancel the units and to be able to compare the values. It showed that birthyear and mass had the highest relative variability (1.57, 1.53) compared to height (0.19). This shows for the first two that their relative spread is big, while for height the values are tightly clustered around the mean (because of the imputation)

* In mass we have Jabba who is very large (1358 kg) and quite old (600 years), but not as old as Yoda with his 896 years.
