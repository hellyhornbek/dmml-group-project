# Data Exploration

```python
import pandas as pd
from pathlib import Path

SAMPLE_PATH = Path("../../Samples/sample_2023_600users.csv")


def print_first_lines(path=SAMPLE_PATH, n=10):
    """Print the first n rows of a CSV file using pandas."""
    df = pd.read_csv(path, nrows=n)
    with pd.option_context("display.max_columns", None, "display.width", None):
        print(df)


print_first_lines()
```
